```python
import re
import pandas as pd
import numpy as np
import urllib.parse

access_pattern = re.compile(
    r'(?P<ip>\S+) \S+ \S+ \[(?P<time>[^\]]+)\] '
    r'"(?P<method>\S+)?\s?(?P<path>\S+)?\s?(?P<protocol>[^"]+)?" '
    r'(?P<status>\d{3}) (?P<bytes>\S+) '
    r'"(?P<referer>[^"]*)" "(?P<user_agent>[^"]*)"'
)

access_data = []
with open('access_log.txt', 'r', encoding='utf-8', errors='ignore') as f:
    for line in f:
        m = access_pattern.match(line)
        if m:
            access_data.append(m.groupdict())

df = pd.DataFrame(access_data)
print("=== 컬럼 목록 ===")
print(df.columns.tolist())
print("\n=== 결측치 개수 ===")
print(df.isnull().sum())

print("\n=== 이례적인 Method/Protocol 확인 ===")
unusual = df[~df['method'].isin(['GET','POST','HEAD','PUT','DELETE','OPTIONS'])]
print(f"비표준 method 건수: {len(unusual)}")
print(unusual[['ip','time','method','path','status']].head(10))
```

```python
# 이전엔 포맷을 안 줘서 전부 NaT 됐던 문제가 있었음 → 포맷 명시
df['datetime'] = pd.to_datetime(
    df['time'].str.replace(r'\s*\+.*$', '', regex=True),
    format='%d/%b/%Y:%H:%M:%S',
    errors='coerce'
)
print(f"datetime 파싱 실패(NaT) 건수: {df['datetime'].isna().sum()} / {len(df)}")

df['day_name']     = df['datetime'].dt.day_name()
df['hour']         = df['datetime'].dt.hour
df['status_group'] = df['status'].astype(str).str[0] + 'xx'

known_bots   = 'CensysInspect|zgrab|nmap|nikto|sqlmap|masscan'
df['is_bot'] = df['user_agent'].str.contains(known_bots, case=False, na=False)

print(df[['datetime','day_name','hour','status_group','is_bot']].head())
```

```python
print("=== IP별 접속 횟수 Top 10 ===")
print(df['ip'].value_counts().head(10))

print("\n=== 요일별 접속 건수 ===")
print(df['day_name'].value_counts())

print("\n=== 시간대별 접속 건수 ===")
print(df['hour'].value_counts().sort_index())

print("\n=== 상태 코드 그룹별 비율 ===")
print(df['status_group'].value_counts(normalize=True) * 100)

df_4xx = df[df['status_group'] == '4xx']
print("\n=== 4xx 에러 최다 발생 IP Top 5 ===")
print(df_4xx['ip'].value_counts().head(5))
```

```python
import bisect, socket, struct

def ip_to_int(ip):
    try:
        return struct.unpack("!I", socket.inet_aton(ip))[0]
    except OSError:
        return None

df_country_raw = pd.read_csv('ipv4.csv', encoding='cp949')

# 컬럼명 실제 이름으로 교체: 국가코드 / 시작IP / 끝IP
df_country_raw['start_int'] = df_country_raw['시작IP'].apply(ip_to_int)
df_country_raw['end_int']   = df_country_raw['끝IP'].apply(ip_to_int)
df_country_raw = (df_country_raw
                   .dropna(subset=['start_int','end_int'])
                   .sort_values('start_int')
                   .reset_index(drop=True))

starts = df_country_raw['start_int'].tolist()

def find_country(ip):
    ip_int = ip_to_int(ip)
    if ip_int is None:
        return 'Unknown'
    idx = bisect.bisect_right(starts, ip_int) - 1
    if idx < 0:
        return 'Unknown'
    row = df_country_raw.iloc[idx]
    return row['국가코드'] if row['start_int'] <= ip_int <= row['end_int'] else 'Unknown'

ip_country_map = pd.DataFrame({'ip': df['ip'].unique()})
ip_country_map['country'] = ip_country_map['ip'].apply(find_country)
ip_country_map.to_csv('ip_country_map.csv', index=False, encoding='utf-8-sig')

df = df.merge(ip_country_map, on='ip', how='left')

print("=== 국가별 상태그룹 분포 ===")
print(df.groupby('country')['status_group'].value_counts().unstack().fillna(0))
```

```python
non_standard = df[~df['method'].isin(['GET','POST','HEAD'])]
print(f"비표준 Method 요청 건수: {len(non_standard)}")
print(non_standard['method'].value_counts())

df['path_decoded'] = df['path'].fillna('').apply(lambda x: urllib.parse.unquote(str(x)))

# '--' 단독은 오탐이 많아 제외, SELECT는 공백 동반 조건으로 좁힘
attack_patterns = r"UNION|SELECT\s|<script>|javascript:|\.\./|/etc/passwd|' OR '1'='1|1=1"
df_attacks = df[df['path_decoded'].str.contains(attack_patterns, case=False, na=False)]

print(f"\n=== 공격 의심 요청 (총 {len(df_attacks)}건) ===")
print(df_attacks[['ip','datetime','method','path_decoded','status']].head(15))
```

```python
error_pattern = re.compile(
    r'\[(?P<time>[^\]]+)\] \[(?:(?P<module>\S+):)?(?P<level>\S+)\] '
    r'\[(?:pid (?P<pid>\d+)(?::tid (?P<tid>\d+))?)\] '
    r'(?:\[client (?P<ip>[^:]+)(?::(?P<port>\d+))?\] )?(?P<message>.*)'
)

error_data = []
with open('error_log.txt', 'r', encoding='utf-8', errors='ignore') as f:
    for line in f:
        m = error_pattern.match(line)
        if m:
            error_data.append(m.groupdict())

df_error = pd.DataFrame(error_data)
df_error['datetime'] = pd.to_datetime(df_error['time'], errors='coerce')
df_error['ip'] = df_error['ip'].astype(str).str.split(':').str[0]

df_access_clean = df.dropna(subset=['datetime']).sort_values('datetime')
df_error_clean  = df_error.dropna(subset=['datetime','ip']).sort_values('datetime')

merged = pd.merge_asof(
    df_access_clean, df_error_clean,
    on='datetime', by='ip', direction='nearest',
    tolerance=pd.Timedelta('60s'), suffixes=('', '_err')
)
error_mapped = merged.dropna(subset=['message'])
print(f"매핑 성공 건수: {len(error_mapped)}")

# 디렉토리 인덱스 경고(AH01276) 등 일상성 로그는 제외하고 공격 관련성 높은 것만 추림
routine_codes = ['AH01276']
attack_related_errors = error_mapped[
    ~error_mapped['message'].str.contains('|'.join(routine_codes), na=False)
]
print(f"공격 관련성 높은 에러 매핑 건수: {len(attack_related_errors)}")
print(attack_related_errors[['ip','datetime','path_decoded','status','level','message']].head(10))
```

```python
import sqlite3

conn = sqlite3.connect('syslog_study.db')

df.to_sql('homepage_data', conn, if_exists='replace', index=False)
df_error.to_sql('error_data', conn, if_exists='replace', index=False)
attack_related_errors.to_sql('mapped_result', conn, if_exists='replace', index=False)

print("테이블 목록:", pd.read_sql("SELECT name FROM sqlite_master WHERE type='table';", conn))
```

```python
query_4xx = """
SELECT ip, MIN(datetime) AS first_4xx_time, COUNT(*) AS cnt
FROM homepage_data
WHERE status LIKE '4%'
GROUP BY ip
ORDER BY first_4xx_time;
"""
print(pd.read_sql(query_4xx, conn))
```
