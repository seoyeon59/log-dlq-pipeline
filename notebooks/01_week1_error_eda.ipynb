```python
from pathlib import Path
import re
import pandas as pd
```

```python
## 0-1 / 경로 설정 ##
pd.set_option("display.max_colwidth", 150)

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
RAW_ERROR     = ROOT / "data" / "raw" / "error.log"
ERROR_PARSED  = ROOT / "data" / "interim" / "error_parsed.csv"
ERROR_FAIL    = ROOT / "data" / "output" / "error_parse_failure.csv"
OUT = ROOT / "reports" / "tables"
OUT.mkdir(parents=True, exist_ok=True)

for p in [RAW_ERROR, ERROR_PARSED, ERROR_FAIL]:
    print(p.exists(), p)
```

```python
## 0-2 / 데이터 로드 + 컬럼 확인 ##

df = pd.read_csv(ERROR_PARSED)
fail_csv = pd.read_csv(ERROR_FAIL)

with open(RAW_ERROR, encoding="utf-8", errors="replace") as f:
    lines = f.read().splitlines()

print("원본 행 수:", len(lines))          # 629 예상
print("error_parsed:", df.shape)
print("parse_failure:", fail_csv.shape)   # 18행 예상
print(df.columns.tolist())
df.head()
```

```python
# time → timestamp로 이름만 바꾸고, 나머지는 이름이 이미 맞음
df = df.rename(columns={"time": "timestamp"})

# timestamp 날짜 타입 변환: time 값 우선, 실패하면 time_raw(Apache 원문 형식)로 보완
t1 = pd.to_datetime(df["timestamp"], errors="coerce")
t2 = pd.to_datetime(df["time_raw"], format="%a %b %d %H:%M:%S.%f %Y", errors="coerce")
df["timestamp"] = t1.fillna(t2)

# is_standard_format이 CSV에서 문자열로 읽혔을 수도 있어서 bool로 통일
df["is_standard_format"] = df["is_standard_format"].astype(str).str.lower().eq("true")

# 분석 기준: 표준 헤더 형식인 행
ok = df[df["is_standard_format"]].copy()

print("전체:", len(df))                                  # 629 예상
print("표준 형식(분석 대상):", len(ok))                     # 611 예상
print("비표준 형식:", int((~df["is_standard_format"]).sum()))  # 18 예상
print("timestamp 변환 실패(표준 형식 중):", int(ok["timestamp"].isna().sum()))   # 0 예상
```

```python
## 원본에서 헤더 없는 행 찾기 ##
raw = pd.DataFrame({"line_no": range(1, len(lines) + 1), "raw": lines})
raw["module"] = raw["raw"].str.extract(r"^\[[^\]]+\]\s+\[([^:\]]*):")[0]
raw["has_header"] = raw["module"].notna()

no_ts = raw[~raw["has_header"]]
print("헤더 없는 행:", len(no_ts), "/ parse_failure.csv:", len(fail_csv))
display(no_ts["raw"].value_counts().to_frame("건수"))
```

```python
## 연속 줄 패턴 여부 ##
cont = no_ts["raw"].str.match(r"^(\s+|at\s|Traceback|Caused by)")
print("연속 줄 패턴:", int(cont.sum()), "/", len(no_ts))
```

```python
## 직전/직후 로그 비교 ##
rows = []
for ln in no_ts["line_no"]:
    prev = raw[raw["line_no"] == ln - 1]
    nxt  = raw[raw["line_no"] == ln + 1]
    next5 = raw[raw["line_no"].between(ln + 1, ln + 5)]["module"].dropna().tolist()
    rows.append({
        "line_no": ln,
        "prev_module": prev["module"].iloc[0] if len(prev) else None,
        "prev_raw": prev["raw"].iloc[0][:90] if len(prev) else None,
        "next_module": nxt["module"].iloc[0] if len(nxt) else None,
        "next_raw": nxt["raw"].iloc[0][:90] if len(nxt) else None,
        "next5_modules": ",".join(next5),
    })
neighbors = pd.DataFrame(rows)
display(neighbors)
print("직후 module 분포:"); display(neighbors["next_module"].value_counts(dropna=False))
print("직전 module 분포:"); display(neighbors["prev_module"].value_counts(dropna=False))
```

```python
## 재시작 가설 확인 + 저장 ##
restart = ["suexec", "lbmethod_heartbeat", "http2", "mpm_event", "core"]
cnt = ok["module"].value_counts().reindex(restart).fillna(0).astype(int)
display(pd.DataFrame({"건수": cnt, "18과 동일": cnt == len(no_ts)}))

neighbors.to_csv(OUT / "error_missing_ts_neighbors.csv", index=False, encoding="utf-8-sig")
```

```python
## Error timestamp 최소/최대/수집 기간 ##

ts = ok["timestamp"].dropna()
ts_summary = pd.DataFrame({"값": {
    "유효 timestamp 건수": len(ts),
    "최소": ts.min(),
    "최대": ts.max(),
    "수집 기간": ts.max() - ts.min(),
    "시간 역전 건수(파일 순서)": int((ts.diff() < pd.Timedelta(0)).sum()),
}})
display(ts_summary)
display(ts.dt.date.value_counts().sort_index().to_frame("일자별 건수"))
ts_summary.to_csv(OUT / "error_timestamp_range.csv", encoding="utf-8-sig")
```

```python
## Level 분포 ##

level_dist = ok["level"].value_counts().to_frame("건수")
level_dist["비율(%)"] = (level_dist["건수"] / len(ok) * 100).round(1)
display(level_dist)

ALLOWED = {"emerg","alert","crit","error","warn","notice","info","debug"} | {f"trace{i}" for i in range(1, 9)}
bad_level = ~ok["level"].isin(ALLOWED)
print("허용값 외 level:", int(bad_level.sum()))
level_dist.to_csv(OUT / "error_level_dist.csv", encoding="utf-8-sig")
```

```python
## Module 분포 ##

module_dist = ok["module"].value_counts().to_frame("건수")
module_dist["비율(%)"] = (module_dist["건수"] / len(ok) * 100).round(1)
display(module_dist)
display(pd.crosstab(ok["module"], ok["level"], margins=True))
module_dist.to_csv(OUT / "error_module_dist.csv", encoding="utf-8-sig")
```

```python
## 코드 별 빈도 ##

code_freq = (ok.groupby(ok["error_code"].fillna("(코드 없음)"))
               .agg(건수=("level", "size"),
                    module=("module", lambda x: ",".join(sorted(x.dropna().unique()))),
                    level=("level",  lambda x: ",".join(sorted(x.dropna().unique()))),
                    예시=("message", "first"))
               .sort_values("건수", ascending=False))
code_freq["비율(%)"] = (code_freq["건수"] / len(ok) * 100).round(1)
print("고유 코드 수:", ok["error_code"].nunique())
display(code_freq)
code_freq.to_csv(OUT / "error_code_freq.csv", encoding="utf-8-sig")
```

```python
## 코드-level 일관성 ##

lv = ok.dropna(subset=["error_code"]).groupby("error_code")["level"].nunique()
mixed_codes = lv[lv > 1].index.tolist()
print("level이 섞인 코드:", mixed_codes)
print("AH 코드 없는 행:", int(ok["error_code"].isna().sum()))
```

```python
## Client IP 등장 빈도 ##

import ipaddress

def ip_type(ip):
    if pd.isna(ip): return None
    try: a = ipaddress.ip_address(str(ip))
    except ValueError: return "INVALID"
    return "LOOPBACK" if a.is_loopback else ("PRIVATE" if a.is_private else "PUBLIC")

ok["client_ip_type"] = ok["client_ip"].map(ip_type)
has_ip = ok["client_ip"].notna()
print(f"client 있음: {has_ip.sum()} / 없음(서버 이벤트): {(~has_ip).sum()}")   # 449 / 162 예상
print("고유 IP:", ok["client_ip"].nunique())
display(ok["client_ip_type"].value_counts(dropna=False).to_frame("건수"))

ip_freq = ok["client_ip"].value_counts().to_frame("건수")
ip_freq["비율(%)"] = (ip_freq["건수"] / has_ip.sum() * 100).round(1)
display(ip_freq.head(15))
ip_freq.to_csv(OUT / "error_client_ip_freq.csv", encoding="utf-8-sig")
```

```python
## 6-2 ##
SCAN_CODES    = ["AH01092", "AH00126", "AH01276"]
BACKEND_CODES = ["AH00957", "AH01114", "AH01095", "AH01102", "AH00898"]
RESTART_CODES = ["AH01232", "AH02282", "AH02951", "AH00489", "AH00094", "AH00492"]

top_ips = ok["client_ip"].value_counts().head(10).index
ip_code = pd.crosstab(ok.loc[ok["client_ip"].isin(top_ips), "client_ip"],
                      ok.loc[ok["client_ip"].isin(top_ips), "error_code"])
display(ip_code)
scan_share = ok.loc[ok["client_ip"].isin(top_ips), "error_code"].isin(SCAN_CODES).mean() * 100
print(f"상위 10개 IP 요청 중 스캐닝 코드 비율: {scan_share:.1f}%")

# 백엔드 IP가 client로 찍힌 5건 확인
display(ok.loc[ok["client_ip"] == "114.70.38.81", ["timestamp", "error_code", "message"]])
ip_code.to_csv(OUT / "error_top_ip_by_code.csv", encoding="utf-8-sig")
```

```python
## 6-3 ##
a = ok.loc[ok["error_code"] == "AH00957", ["timestamp"]].sort_values("timestamp").rename(columns={"timestamp": "ts_957"})
b = ok.loc[ok["error_code"] == "AH01114", ["timestamp"]].sort_values("timestamp")
pair = pd.merge_asof(b, a, left_on="timestamp", right_on="ts_957",
                     direction="nearest", tolerance=pd.Timedelta("1s"))
n_pair = int(pair["ts_957"].notna().sum())
print(f"AH01114 {len(b)}건 중 1초 이내에 AH00957이 짝으로 있는 건: {n_pair}건")
print("최대 시간 차:", (pair["timestamp"] - pair["ts_957"]).abs().max())
```

```python
## 6-4 ##
has_client = ok["client_ip"].notna().map({True: "client 있음", False: "client 없음"})
code_client = pd.crosstab(ok["error_code"], has_client)
display(code_client)

mixed_client_codes = code_client[(code_client.get("client 있음", 0) > 0) &
                                 (code_client.get("client 없음", 0) > 0)].index.tolist()
e12_mask = ok["error_code"].isin(mixed_client_codes) & ok["client_ip"].isna()
print("client 유무가 섞인 코드:", mixed_client_codes)
print("그중 client 없는 행:", int(e12_mask.sum()))
code_client.to_csv(OUT / "error_code_by_client.csv", encoding="utf-8-sig")
```

```python
## 6-5 ##
all_days = pd.date_range(ts.min().normalize(), ts.max().normalize(), freq="D").date
missing_days = sorted(set(all_days) - set(ts.dt.date))
print(f"전체 {len(all_days)}일 중 로그 없는 날: {len(missing_days)}일")
print([str(d) for d in missing_days])
```

```python
## 7-2 / Error 품질 규칙 후보 ##

N = len(lines)
rules = [
    ("E01", "표준 헤더 없음(timestamp 누락)", len(no_ts),                                           "행", N, "형식 결함",   "데이터 품질", "VALID_ANOMALY 후보"),
    ("E02", "timestamp 변환 실패",           int(ok["timestamp"].isna().sum()),                     "행", N, "형식 결함",   "데이터 품질", "INVALID_DATA 후보"),
    ("E03", "허용값 외 level",               int(bad_level.sum()),                                  "행", N, "형식 결함",   "데이터 품질", "VALID_ANOMALY 후보"),
    ("E04", "module 누락",                   int(ok["module"].isna().sum()),                        "행", N, "형식 결함",   "데이터 품질", "INVALID_DATA 후보"),
    ("E05", "AH 코드 없음",                  int(ok["error_code"].isna().sum()),                    "행", N, "형식 결함",   "데이터 품질", "VALID_ANOMALY 후보"),
    ("E06", "같은 코드의 level 불일치",       int(ok["error_code"].isin(mixed_codes).sum()),         "행", N, "일관성",      "데이터 품질", "VALID_ANOMALY 후보"),
    ("E07", "client IP 형식 오류",           int((ok["client_ip_type"] == "INVALID").sum()),        "행", N, "형식 결함",   "데이터 품질", "INVALID_DATA 후보"),
    ("E08", "시간 역전",                     int((ok["timestamp"].diff() < pd.Timedelta(0)).sum()), "행", N, "일관성",      "데이터 품질", "VALID_ANOMALY 후보"),
    ("E09", "스캐닝·공격성 요청",             int(ok["error_code"].isin(SCAN_CODES).sum()),          "행", N, "보안 이벤트", "내용 이상",   "VALID_ANOMALY 후보"),
    ("E10", "동일 장애 중복 기록(AH00957+AH01114)", n_pair,                                          "쌍", len(b), "집계 주의", "내용 이상", "VALID (집계 시 중복 제거)"),
    ("E11", "서버 재시작 시퀀스 이벤트",       int(ok["error_code"].isin(RESTART_CODES).sum()),       "행", N, "운영 이벤트", "내용 이상",   "VALID (태그만)"),
    ("E12", "같은 코드인데 client 정보 누락",  int(e12_mask.sum()),                                   "행", N, "완전성",      "데이터 품질", "VALID_ANOMALY 후보"),
    ("E13", "로그가 없는 날짜",               len(missing_days),                                     "일", len(all_days), "적시성", "데이터 품질", "확인 필요 (Access와 대조)"),
]
rule_table = pd.DataFrame(rules, columns=["규칙 ID", "규칙", "위반 수", "단위", "분모", "이상 유형", "구분", "잠정 분류"])
rule_table["비율(%)"] = (rule_table["위반 수"] / rule_table["분모"] * 100).round(2)
display(rule_table)
rule_table.to_csv(OUT / "error_rule_candidates.csv", index=False, encoding="utf-8-sig")
```

```python
## 8-1 / 행마다 태그 달고 분류 ##

def row_tags(r):
    tags = []
    if r["error_code"] in SCAN_CODES:    tags.append("E09")
    if r["error_code"] in RESTART_CODES: tags.append("E11")
    if r["error_code"] in BACKEND_CODES: tags.append("BACKEND")
    if r["_e12"]:                        tags.append("E12")
    return ",".join(tags)

ok["_e12"] = e12_mask
ok["tags"] = ok.apply(row_tags, axis=1)

def classify(tags):
    t = set(tags.split(",")) if tags else set()
    return "VALID_ANOMALY" if t & {"E09", "E12"} else "VALID"
ok["quality_class"] = ok["tags"].map(classify)

def event_type(tags):
    if "E09" in tags: return "보안(스캐닝)"
    if "E11" in tags: return "운영(재시작)"
    if "BACKEND" in tags: return "운영(백엔드 장애)"
    return "기타"
ok["event_type"] = ok["tags"].map(event_type)

# 헤더 없는 18행도 포함해서 전체 629행 기준으로 합치기
hdr = pd.DataFrame({"line_no": no_ts["line_no"].values, "error_code": "AH00558",
                    "tags": "E01", "quality_class": "VALID_ANOMALY", "event_type": "운영(재시작)"})
row_class = pd.concat([ok.drop(columns="_e12"), hdr], ignore_index=True)

display(pd.crosstab(row_class["event_type"], row_class["quality_class"], margins=True, margins_name="합계"))
qc = row_class["quality_class"].value_counts().to_frame("건수")
qc["비율(%)"] = (qc["건수"] / len(row_class) * 100).round(1)
display(qc)

row_class.to_csv(OUT / "error_row_classification.csv", index=False, encoding="utf-8-sig")
```

```python
## 8-2 / valid 정의표 ##

def ids(cond): return ", ".join(rule_table.loc[cond, "규칙 ID"])

valid_def = pd.DataFrame([
    ("VALID",
     "표준 헤더 존재, 필수 필드 추출 성공, 조사 대상 규칙 위반 없음. 재시작·백엔드 장애처럼 정상 기록된 운영 이벤트 포함(태그로 구분)",
     ids(rule_table["잠정 분류"].str.startswith("VALID (")),
     int((row_class["quality_class"] == "VALID").sum())),
    ("VALID_ANOMALY",
     "형식은 읽히거나 복구 가능하지만 조사가 필요한 데이터. 데이터 품질 이상(E01, E12)과 내용 이상(E09 스캐닝)으로 나뉨",
     ids(rule_table["잠정 분류"].str.startswith("VALID_ANOMALY")),
     int((row_class["quality_class"] == "VALID_ANOMALY").sum())),
    ("INVALID_DATA",
     "분석에 필요한 핵심 정보를 복구할 수 없는 데이터",
     ids(rule_table["잠정 분류"].str.startswith("INVALID")),
     int((row_class["quality_class"] == "INVALID_DATA").sum())),
], columns=["분류", "정의", "해당 규칙", "현재 건수"])
display(valid_def)
valid_def.to_csv(OUT / "error_valid_definition_draft.csv", index=False, encoding="utf-8-sig")
```

```python
## 품질 탐색 결과표 통합 ##

lv = ok["level"].value_counts()
scan_in_warn = ok.loc[ok["level"] == "warn", "error_code"].isin(SCAN_CODES).mean() * 100
by_date_top = ts.dt.date.value_counts().head(2)
restart_days = ok.loc[ok["error_code"] == "AH00489", "timestamp"].dt.date.value_counts().sort_index()

findings = pd.DataFrame([
    ("Error", "수집 기간",
     f"{ts.min():%Y-%m-%d %H:%M} ~ {ts.max():%Y-%m-%d %H:%M} ({(ts.max()-ts.min()).days}일)",
     f"{len(all_days)}일 중 {len(missing_days)}일 로그 없음 / " + ", ".join(f"{d} {n}건" for d, n in by_date_top.items()) + " 집중",
     "E08, E13"),
    ("Error", "헤더 없는 행", f"{len(no_ts)}건 (모두 AH00558)",
     "재시작 시퀀스 내 독립 메시지, 앞뒤 행으로 시각 복구 가능 / ServerName 미설정 반복", "E01"),
    ("Error", "Level 분포", ", ".join(f"{k} {v}" for k, v in lv.items()),
     f"warn의 {scan_in_warn:.0f}%가 스캐닝성 요청", "E03, E09"),
    ("Error", "Module 분포", ", ".join(f"{k} {v}" for k, v in ok["module"].value_counts().head(5).items()),
     f"재시작 관련 module 18건씩 → 재시작 {len(no_ts)}회 (" + ", ".join(f"{d} {n}회" for d, n in restart_days.items()) + ")",
     "E04, E11"),
    ("Error", "Error code", f"고유 {ok['error_code'].nunique()}개",
     f"스캐닝 {int(ok['error_code'].isin(SCAN_CODES).sum())}건 / 백엔드 장애 {int(ok['error_code'].isin(BACKEND_CODES).sum())}건(중복 쌍 {n_pair}) / 재시작 {int(ok['error_code'].isin(RESTART_CODES).sum())}건",
     "E05, E06, E09, E10, E11"),
    ("Error", "Client IP", f"고유 {ok['client_ip'].nunique()}개 / client 있는 행 {int(ok['client_ip'].notna().sum())}건",
     f"상위 10개 IP의 {scan_share:.0f}%가 스캐닝 코드 / 동일 건수 반복 IP 다수 / client 누락 {int(e12_mask.sum())}건",
     "E07, E12"),
], columns=["로그", "항목", "측정값", "발견 이슈", "관련 규칙"])
display(findings)
findings.to_csv(OUT / "error_quality_findings.csv", index=False, encoding="utf-8-sig")
```
